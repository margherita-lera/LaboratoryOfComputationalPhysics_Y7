## Pandas analysis

This exercise consists in analyzing a dataset containg timing information from a series of Time-to-Digital-Converters (TDC) implemented in a couple of FPGAs. Each measurement (i.e. each row of the input file) consists of a flag that specifies the type of message ('HEAD', which in this case is always 1), two addresses of the TDC providing the signal ('FPGA' and 'TDC_CHANNEL'), and the timing information ('ORBIT_CNT', 'BX_COUNTER', and 'TDC_MEAS'). Each TDC count corresponds to 25/30 ns, whereas a unit of BX_COUNTER corresponds to 25 ns, and the ORBIT_CNT is increased every 'x' BX_COUNTER. This allows to store the time in a similar way to hours, minutes and seconds.

1\. Create a Pandas DataFrame reading N rows of the 'data_000637.txt' dataset. Choose N to be smaller than or equal to the maximum number of rows and larger that 10k.

2\. Find out the number of BX in a ORBIT (the value 'x').



```python
file_name="data/data_000637.txt"
data=pd.read_csv(file_name)

dfa= data.sample(random.randrange(10001,data.shape[0]),axis=0)

#DISTRIBUTION
#plt.hist(dfa['BX_COUNTER'])   #I did this to check if we have 
# enough data to be sure of the number.

x=max(dfa['BX_COUNTER'])+1
print('x: ',x)
```

```python
display(dfa.tail(10))
print(dfa.shape[0])
```

3\. Find out how much the data taking lasted. You can either make an estimate based on the fraction of the measurements (rows) you read, or perform this check precisely by reading out the whole dataset.




```python
#I find the row with the maximum and minimum time, going from orbit 
#count to tdc_meas if needed, until I find only 1 row.
#max time != last row! -->
#N=data.shape[0]
#print('N: ',N)
#print(data.iloc[N-1])
#data.tail(5)

max_orbit= data[data['ORBIT_CNT']==data['ORBIT_CNT'].max()]
s2=max_orbit[max_orbit['BX_COUNTER']==max_orbit['BX_COUNTER'].max()]
print('s2:\n',s2)

min_orbit= data[data['ORBIT_CNT']==data['ORBIT_CNT'].min()]
s1=min_orbit[min_orbit['BX_COUNTER']==min_orbit['BX_COUNTER'].min()]
s1=s1[s1['TDC_MEAS']==s1['TDC_MEAS'].min()]
print('s1:\n',s1)

time2=(s2['ORBIT_CNT']*x + (s2['TDC_MEAS']/30 + s2['BX_COUNTER'])*25).iloc[0]*10**(-9)
time1=(s1['ORBIT_CNT']*x + (s1['TDC_MEAS']/30 + s1['BX_COUNTER'])*25).iloc[0]*10**(-9)
time_total=(time2-time1)
print('Total time: ',time_total,'seconds')
```


4\. Create a new column with the absolute time in ns (as a combination of the other three columns with timing information).

```python
data['TIME']=data['ORBIT_CNT']*x + (data['TDC_MEAS']/30 + data['BX_COUNTER'])*25
#data.assign(TIME=lambda y: (y['ORBIT_CNT']*x + (y['TDC_MEAS']/30 + y['BX_COUNTER'])*25))
data.head(10)
```

5\. Replace the values (all 1) of the HEAD column randomly with 0 or 1.


```python
data['HEAD']=[random.randrange(0,2,1) for x in range(data.shape[0])]
data.head(10)
```


6\. Create a new DataFrame that contains only the rows with HEAD=1.



```python
data_filtered=data[data['HEAD']==1]
data_filtered.head()
```

7\. Make two occupancy plots (one for each FPGA), i.e. plot the number of counts per TDC channel



```python
#i need to sum orbit cnt for each tbc
plots=data.drop(['HEAD','BX_COUNTER','TDC_MEAS'],axis=1)
plot1=plots[plots['FPGA']==1].drop(['FPGA'],axis=1)
plot1=plot1.groupby('TDC_CHANNEL').sum()

plot0=plots[plots['FPGA']==0].drop(['FPGA'],axis=1)
plot0=plot0.groupby('TDC_CHANNEL').sum()

plot0.tail(10)
plot1.tail(10)
plot1.plot()
plot0.plot()
```

8\. Use the groupby method to find out the noisy channels, i.e. the TDC channels with most counts (say the top 3)


```python
result=data.groupby('TDC_CHANNEL').size().sort_values(ascending=False)

print(result[0:3])
```
